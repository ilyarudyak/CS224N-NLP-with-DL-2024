## 2.1 Modified n-gram precision

```python
# Example 1 
candidate_1 = (
    "It is a guide to action which ensures that the military "
    "always obeys the commands of the party."
)

candidate_2 = (
    "It is to insure the troops forever hearing the activity "
    "guidebook that party direct."
)

references = [
    (
        "It is a guide to action that ensures that the military "
        "will forever heed Party commands."
    ),
    (
        "It is the guiding principle which guarantees the military "
        "forces always being under the command of the Party."
    ),
    (
        "It is the practical guide for the army always to heed "
        "the directions of the party."
    ),
]
```

```python
# Instantiate BleuScoreV1 
bleu = BleuScoreV1()

# Compute modified unigram precision for the candidate sentences against the references
clipped_count, candidate_count, modified_precision = bleu.modified_unigram_precision(candidate_1, references)
print(f"Candidate 1: clipped_count={clipped_count}, candidate_count={candidate_count}, modified_precision={modified_precision:.4f}")
clipped_count, candidate_count, modified_precision = bleu.modified_unigram_precision(candidate_2, references)
print(f"Candidate 2: clipped_count={clipped_count}, candidate_count={candidate_count}, modified_precision={modified_precision:.4f}")
```

```python
# Instantiate BleuScoreV1 
bleu2 = BleuScoreV2()

# Compute modified unigram precision for the candidate sentences against the references
clipped_count, candidate_count, modified_precision = bleu2.modified_bigram_precision(candidate_1, references)
print(f"Candidate 1: clipped_count={clipped_count}, candidate_count={candidate_count}, modified_precision={modified_precision:.4f}")
clipped_count, candidate_count, modified_precision = bleu2.modified_bigram_precision(candidate_2, references)
print(f"Candidate 2: clipped_count={clipped_count}, candidate_count={candidate_count}, modified_precision={modified_precision:.4f}")
```
