```python
import sqlite3
```

```python
# Connect to a file on disk (creates 'geography.db' in the current folder)
conn = sqlite3.connect("geography.db")
cursor = conn.cursor()

# Reset table if it already exists
cursor.execute("DROP TABLE IF EXISTS states")

# Table schema with explicit column names
cursor.execute(
    """
CREATE TABLE states (
    state TEXT PRIMARY KEY,
    borders_texas INTEGER
)
"""
)

# Insert sample data (1 = True, 0 = False)
states_data = [
    ("Texas", 0),  # A state doesn't border itself
    ("Oklahoma", 1),
    ("Louisiana", 1),
    ("New Mexico", 1),
    ("Arkansas", 1),
    ("California", 0),
    ("New York", 0),
]

cursor.executemany("INSERT INTO states VALUES (?, ?)", states_data)
conn.commit()
conn.close()
```

```python
def simple_semantic_parser(sentence: str) -> str:
    # A toy rule/pattern matcher representing what CCG or Seq2Seq does
    cleaned = sentence.lower().strip("?")

    if "states" in cleaned and "border" in cleaned and "texas" in cleaned:
        # Maps the natural intent to an exact formal query
        return "SELECT state FROM states WHERE borders_texas = 1;"
    else:
        raise ValueError("Cannot parse query into logical form.")
```

```python
# User question
user_question = "What states border Texas?"

# Semantic Parsing: parse the user question into a SQL query
sql_query = simple_semantic_parser(user_question)
print(f"Parsed Logical Form: {sql_query}")
```
