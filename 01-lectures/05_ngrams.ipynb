## 1 Building a Basic Language Model

### ========DEBUGGING========

```python
sentence = reuters.sents()[10]
print(sentence)
print(' '.join(sentence))
# extract bigrams and trigrams
print(list(bigrams(sentence)))
print(list(trigrams(sentence)))
```

```python
model = {}
for sentence in reuters.sents()[:10]:
    for w1, w2, w3 in trigrams(sentence, pad_right=True, pad_left=True):
        context = (w1, w2)
        
        # Initialize the outer dictionary
        if context not in model:
            model[context] = {}
        
        # Initialize the inner dictionary's count
        if w3 not in model[context]:
            model[context][w3] = 0
            
        model[context][w3] += 1
```

```python
model
```

```python
for w1_w2 in model:
    total_count = float(sum(model[w1_w2].values()))
    for w3 in model[w1_w2]:
        model[w1_w2][w3] /= total_count
```
