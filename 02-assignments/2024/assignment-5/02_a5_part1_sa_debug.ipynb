## 01 Datasets

### 01 Stanford Sentiment Treebank (SST) dataset

```python
ids_sst_train = pd.read_csv("data/ids-sst-train.csv", sep="\t")
ids_sst_dev = pd.read_csv("data/ids-sst-dev.csv", sep="\t")
ids_sst_test = pd.read_csv("data/ids-sst-test-student.csv", sep="\t")
```

```python
ids_sst_train.shape, ids_sst_dev.shape, ids_sst_test.shape
```

```python
ids_sst_train.head()
```

```python
# Print a few reviews from the ids-sts dataset with a sentiment score
for i in range(3):
    print(f"Review: {ids_sst_train.iloc[i]['sentence'][:100]}...")
    print(f"Sentiment score: {ids_sst_train.iloc[i]['sentiment']}")
    print()
```

```python
# Extract unique sentiment scores from the ids-sts dataset
unique_sentiments = ids_sst_train['sentiment'].unique()
sorted(unique_sentiments.tolist())
```
